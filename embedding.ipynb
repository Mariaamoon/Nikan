```python
from quotes import QUOTES
```

```python
texts = []
labels = []

for q in QUOTES:
    texts.append(q["text"])

    labels.append(q["mood"])
```

```python
print(len(texts))
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords


stop_words = set(stopwords.words('english'))



def preprocess(text):
    try:
        tokens = word_tokenize(text.lower())

        tokens = [
            w for w in tokens
            if w.isalpha()
        ]

        tokens = [
            w for w in tokens
            if w not in stop_words
        ]
    
    except LookupError as e:
        # Catch errors from external resources (NLTK/spaCy/etc.)
        print( f"[lookup_error]: {e}")

    return " ".join(tokens)
```

```python
clean_texts = []


for t in texts:
    clean_texts.append(
        preprocess(t)
    )
```

```python
from sklearn.preprocessing import LabelEncoder


encoder = LabelEncoder()

y = encoder.fit_transform(labels)
```

```python
from keras.preprocessing.sequence import pad_sequences


MAX_LEN = 30


X = pad_sequences(
        clean_texts,
        maxlen=MAX_LEN,
        padding='post'
)
```
