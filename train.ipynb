```python
from quotes import QUOTES
```

```python
texts = []
labels = []

for q in QUOTES:
    texts.append(q["text"])

    labels.append(q["mood"])
```

```python
from sklearn.preprocessing import LabelEncoder


encoder = LabelEncoder()

y = encoder.fit_transform(labels)
```

```python
encoder.classes_
```

```python
from collections import Counter

moods = [quote["mood"] for quote in QUOTES]

mood_counts = Counter(moods)

for mood, count in mood_counts.items():
        print(f"{mood} : {count}")
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords


stop_words = set(stopwords.words('english'))



def preprocess(text):
    try:
        tokens = word_tokenize(text.lower())

        tokens = [
            w for w in tokens
            if w.isalpha()
        ]

        tokens = [
            w for w in tokens
            if w not in stop_words
        ]
    
    except LookupError as e:
        # Catch errors from external resources (NLTK/spaCy/etc.)
        print( f"[lookup_error]: {e}")

    return " ".join(tokens)
```

```python
clean_texts = []


for t in texts:
    clean_texts.append(
        preprocess(t)
    )
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer


vectorizer = TfidfVectorizer()


X = vectorizer.fit_transform(
        clean_texts
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer


vectorizer = TfidfVectorizer()


X = vectorizer.fit_transform(
        clean_texts
)
```

```python
print(X.shape)
```

```python
from sklearn.model_selection import train_test_split


X_train,X_test,y_train,y_test = train_test_split(

X.toarray(),
y,

test_size=0.3
)
```

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import Dropout

model = Sequential()

model.add(
Dense(64,
activation='relu',
input_shape=(X_train.shape[1],))
)


model.add(
Dense(32,
activation='relu')
)
model.add(Dropout(0.3))


model.add(
Dense(
len(set(y)),
activation='softmax'
)
)
```

```python
model.compile(

optimizer='adam',

loss='sparse_categorical_crossentropy',

metrics=['accuracy']

)
```

```python
model.fit(

X_train,
y_train,

epochs=20,

batch_size=32

)
```

```python
loss,acc = model.evaluate(
X_test,
y_test
)

print(acc)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
quote_embeddings = model.encode(texts)
```

```python
user = "I feel hopeless"

user_embedding = model.encode(user)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

scores = cosine_similarity(
    [user_embedding],
    quote_embeddings
)

print(scores)
```

```python
best = scores.argmax()

quote = QUOTES[best]

print(quote["text"])
print(quote["author"])
```
